## Data

```python
data = execute_mysql_command('SELECT * FROM student_data')
df = pd.DataFrame(data, columns=['id', 'Gender', 'Age', 'City', 'Profession', 'Academic_Pressure', 'Work_Pressure', 'CGPA', 'Study_Satisfaction', 'Job_Satisfaction', 'Sleep_Duration', 'Dietary_Habits', 'Degree', 'had_suicidal_thoughts', 'Work_Study_Hours',
'Financial_Stress','Family_History_of_Mental_Illness','Depression'])
df.set_index('id', inplace=True)
df
```

# DATA PREPARATION

## All Columns

```python
df.dtypes
```

```python
print(df.shape)
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
print(df.shape)
```

## Gender

```python
df.Gender.value_counts()
```

0 - male <br>
1 - female

```python
df.Gender = df.Gender.apply(lambda x: 0 if x == 'Male' else 1)
df 
```

## Age

```python
df.Age.dtype
```

```python
df.Age = df.Age.astype(int)
df
```

## City

```python
df_cleaned = df.drop(['City'], axis=1)
df_cleaned
```

## Profession

```python
df_cleaned.Profession.unique()
```

0 - student <br>
1 - other

```python
df_cleaned.Profession = df_cleaned.Profession.apply(
    lambda x: 0 if x == 'Student' else 1)
df_cleaned
```

Delete column because the data is very unbalanced

```python
df_cleaned.Profession.value_counts()
```

```python
df_cleaned.drop(['Profession'], axis=1, inplace= True)
df_cleaned
```

## Sleep duration

```python
df_cleaned.Sleep_Duration.unique()
```

```python
df_cleaned.Sleep_Duration.value_counts()  
```

```python
df_cleaned.query('Sleep_Duration != "Others"', inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(df_cleaned[['Sleep_Duration']])
encoded_df = pd.DataFrame(
    encoded_data, columns=encoder.get_feature_names_out(['Sleep_Duration']))
encoded_df.index = df_cleaned.index
encoded_df
```

```python
df_cleaned = pd.concat([df_cleaned, encoded_df], axis=1)
df_cleaned.drop(['Sleep_Duration'], axis=1, inplace=True)
df_cleaned
```

## Processing numeric columns

```python
from sklearn.preprocessing import StandardScaler
```

```python
df_cleaned['Financial_Stress'].unique()
```

```python
columns_to_process = ['Age', 'Academic_Pressure',
                      'Work_Pressure', 'CGPA', 'Study_Satisfaction', 'Job_Satisfaction', 'Work_Study_Hours', 'Financial_Stress']

df_cleaned.query('Financial_Stress != "?"', inplace=True)
df_cleaned[columns_to_process] = df_cleaned[columns_to_process].astype(float)

df_cleaned['Financial_Stress'].unique()
```

```python
scaler = StandardScaler()
df_num_processed = pd.DataFrame(scaler.fit_transform(df_cleaned[columns_to_process]), columns= columns_to_process)
df_num_processed.index = df_cleaned.index
df_cleaned[columns_to_process] = df_num_processed[columns_to_process]
df_cleaned
```

## Dietary habits

```python
df_cleaned['Dietary_Habits'].value_counts()
```

```python
df_cleaned.query('Dietary_Habits != "Others"', inplace=True)
print(df_cleaned.shape)
df_cleaned['Dietary_Habits'].value_counts()
```

```python
encoder_data_dietary = encoder.fit_transform(df_cleaned[['Dietary_Habits']])
encoder_df_dietary = pd.DataFrame(encoder_data_dietary,
                          columns=encoder.get_feature_names_out())
encoder_df_dietary.index = df_cleaned.index
df_cleaned = pd.concat([df_cleaned, encoder_df_dietary], axis=1)
df_cleaned.drop(['Dietary_Habits'], axis = 1, inplace=True)
```

## Degree

```python
df_cleaned.Degree.value_counts()
```

```python
df_cleaned.query('Degree != "Others"', inplace=True)
df_cleaned['Degree'].value_counts()
```

```python
encoder_data_degree = encoder.fit_transform(df_cleaned[['Degree']])
encoder_df_degree = pd.DataFrame(encoder_data_degree,
                                  columns=encoder.get_feature_names_out())
encoder_df_degree.index = df_cleaned.index
df_cleaned = pd.concat([df_cleaned, encoder_df_degree], axis=1)
df_cleaned.drop(['Degree'], axis=1, inplace=True)
df_cleaned
```

## Had suicidal thoughts

```python
df_cleaned['had_suicidal_thoughts'].value_counts()
```

0 - No <br>
1 - Yes

```python
df_cleaned['had_suicidal_thoughts'] = df_cleaned['had_suicidal_thoughts'].apply(lambda x: 1 if x == 'Yes' else 0)
df_cleaned
```

## Family history of mental illness

```python
df_cleaned['Family_History_of_Mental_Illness'].value_counts()
```

0 - No <br>
1 - Yes

```python
df_cleaned['Family_History_of_Mental_Illness'] = df_cleaned['Family_History_of_Mental_Illness'].apply(
    lambda x: 1 if x == 'Yes' else 0)
df_cleaned
```
